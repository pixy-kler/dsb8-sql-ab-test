* 1. Create a connection to the database using the sqlite3 library.

```python
conn = sqlite3.connect('data/checking-logs.sqlite')
```

* 2. Get the schema of the pageviews table using pd.io.sql.read_sql() and the query "PRAGMA table_info(pageviews);"

```python
pd.io.sql.read_sql("PRAGMA table_info(pageviews);", conn)
```

* 3. Get only the first ten rows of the pageviews table to see what the table looks like.

```python
first_ten = pd.read_sql_query("SELECT * FROM pageviews LIMIT 10;", conn)
print(first_ten)
```

* 4. Use one query to get the subtable where:

    - Only "uid" and "datetime" are used.
    - Only user data (user_*), not admin data, is used.
    - It is sorted by "uid" in ascending order.
    - The index column is "datetime".
    - "datetime" is converted to a DatetimeIndex.
    - The name of the dataframe is pageviews.


```python
pageviews = pd.read_sql_query(
    """
    SELECT uid, datetime
    FROM pageviews
    WHERE uid LIKE 'user_%'
    ORDER BY uid ASC
    """,
    conn,
    index_col='datetime',
    parse_dates=['datetime']
)
```
